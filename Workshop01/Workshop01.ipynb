```python
# Task 4 - Loops and functions
# (a) Print 1 through 10. The stop value of range is exclusive.
for number in range(1, 11):
    print(number, end=" ")
print()

# (b) Start at 2 and step by 2 to print even numbers.
for number in range(2, 11, 2):
    print(number, end=" ")
print()

# (c) Each new Fibonacci number is the sum of the previous two.
current, following = 0, 1
count = 0
while count < 5:
    print(current, end=" ")
    current, following = following, current + following
    count += 1
print()

# (d) Return the sum of two numbers.
def add_numbers(first, second):
    return first + second

# (e) Return the largest element of a non-empty list.
def largest_number(numbers):
    largest = numbers[0]
    for number in numbers:
        if number > largest:
            largest = number
    return largest

# (f) Count vowels regardless of capitalisation.
def count_vowels(text):
    count = 0
    for letter in text.lower():
        if letter in "aeiou":
            count += 1
    return count

print("Sum:", add_numbers(3, 4))
print("Largest:", largest_number([2, 9, 5]))
print("Vowels:", count_vowels("Elephant"))
```

```python
# Task 5 - Caesar cipher
def caesar_cipher(message, shift):
    lowercase = "abcdefghijklmnopqrstuvwxyz"
    uppercase = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    result = ""

    for letter in message:
        if letter in lowercase:
            alphabet = lowercase
        elif letter in uppercase:
            alphabet = uppercase
        else:
            result += letter
            continue

        # Find the letter and wrap around the alphabet after shifting.
        for position in range(26):
            if letter == alphabet[position]:
                new_position = (position + shift) % 26
                result += alphabet[new_position]
                break
    return result

message = input("Enter a message (e.g. HELLO): ")
key = input("Enter a shift from 0 to 25 (e.g. 3): ")
mode = input("Type encrypt or decrypt: ").lower()

valid_key = key != ""
for digit in key:
    if digit not in "0123456789":
        valid_key = False

if not valid_key:
    print("Please run the cell again and enter a shift from 0 to 25.")
elif int(key) > 25:
    print("Please run the cell again and enter a shift from 0 to 25.")
elif mode == "encrypt":
    print("Result:", caesar_cipher(message, int(key)))
elif mode == "decrypt":
    print("Result:", caesar_cipher(message, -int(key)))
else:
    print("Please run the cell again and type encrypt or decrypt.")
```

```python
# Task 6 - Hangman
import random

words = ["python", "java", "javascript"]
word = random.choice(words)
guessed_letters = []
wrong_guesses = 0
max_guesses = 6

while wrong_guesses < max_guesses:
    display_word = ""
    letters_left = 0

    # Show guessed letters and hide the others.
    for letter in word:
        if letter in guessed_letters:
            display_word += letter + " "
        else:
            display_word += "_ "
            letters_left += 1
    print("Word:", display_word)

    if letters_left == 0:
        print("You won! The word was", word)
        break

    print("Wrong guesses:", wrong_guesses, "/", max_guesses)
    guess = input("Guess one English letter: ").lower()

    if len(guess) != 1 or guess not in "abcdefghijklmnopqrstuvwxyz":
        print("Enter one English letter.")
    elif guess in guessed_letters:
        print("Already guessed.")
    else:
        guessed_letters.append(guess)
        if guess not in word:
            wrong_guesses += 1

if wrong_guesses == max_guesses:
    print("Game over. The word was", word)
```
