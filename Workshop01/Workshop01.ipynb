```python
# Task 4 - Loops and functions
# (a) Print 1 through 10. The stop value of range is exclusive.
for number in range(1, 11):
    print(number, end=" ")
print()

# (b) Start at 2 and step by 2 to print even numbers.
for number in range(2, 11, 2):
    print(number, end=" ")
print()

# (c) Each new Fibonacci number is the sum of the previous two.
current, following = 0, 1
count = 0
while count < 5:
    print(current, end=" ")
    current, following = following, current + following
    count += 1
print()

# (d) Return the sum of two numbers.
def add_numbers(first, second):
    return first + second

# (e) Return the largest element of a non-empty list.
def largest_number(numbers):
    return max(numbers)

# (f) Count vowels regardless of capitalisation.
def count_vowels(text):
    return sum(letter in "aeiou" for letter in text.lower())

print("Sum:", add_numbers(3, 4))
print("Largest:", largest_number([2, 9, 5]))
print("Vowels:", count_vowels("Elephant"))
```

```python
# Task 5 - Caesar cipher
def caesar_cipher(message, key, mode):
    # Decryption reverses the shift used for encryption.
    if mode not in ("encrypt", "decrypt"):
        raise ValueError("Choose encrypt or decrypt.")
    shift = key if mode == "encrypt" else -key
    result = []

    for letter in message:
        if "A" <= letter <= "Z":
            base = ord("A")
        elif "a" <= letter <= "z":
            base = ord("a")
        else:
            result.append(letter)
            continue
        # Modulo wraps the new position around the alphabet.
        result.append(chr(base + (ord(letter) - base + shift) % 26))

    return "".join(result)

message = input("Enter a message using English letters (e.g. HELLO): ")

# Ask again until the shift is a whole number.
while True:
    try:
        key = int(input("Enter a whole-number shift (e.g. 3): "))
        break
    except ValueError:
        print("The shift must be a whole number. Try 3.")

mode = input("Type encrypt to encode or decrypt to decode: ").strip().lower()
while mode not in ("encrypt", "decrypt"):
    mode = input("Please type encrypt or decrypt: ").strip().lower()

print("Result:", caesar_cipher(message, key, mode))

encrypted = caesar_cipher("HELLO", 3, "encrypt")
print("Decryption example:", caesar_cipher(encrypted, 3, "decrypt"))
```

```python
# Task 6 - Hangman
from random import choice

def play_hangman(word, remaining):
    guessed = set()

    while remaining > 0:
        # Reveal all positions of letters the player has guessed.
        print(" ".join(letter if letter in guessed else "_" for letter in word))
        if all(letter in guessed for letter in word):
            print(f"You won! The word was {word}.")
            return

        guess = input(f"Guess a letter ({remaining} wrong guesses left): ").strip().lower()
        if len(guess) != 1 or not ("a" <= guess <= "z"):
            print("Enter one English letter.")
            continue
        if guess in guessed:
            print("Already guessed.")
            continue

        guessed.add(guess)
        if guess not in word:
            remaining -= 1

    print(f"Game over. The word was {word}.")

# Choose a random word and give the player six wrong guesses.
words = ["python", "java", "javascript"]
play_hangman(choice(words), 6)
```
