```python
# Task4 (e)
# Write a function that takes a list of numbers as a parameter and returns the largest 
# number in the list.
def find_largest_number(numbers):
    largest = numbers[0]
    for number in numbers:
        if number > largest:
            largest = number
    return largest
print (find_largest_number([3, 5, 2, 8, 1]))
```

```python
# Task4 (f)
# Write a function that takes a string as a parameter and returns the number of vowels in the 
# string.

def count_vowels(string):
    vowels = "AEIOUaeiou"
    count = 0
    for char in string:
        if char in vowels:
            count += 1
    return count
print (count_vowels("Hello world"))
```

```python
# Task 5 - Caesar cipher
def cipher_ceaza(n, word):
    lowercase = "abcdefghijklmnopqrstuvwxyz"
    uppercase = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    new_word = ""

    for letter in word:
        if letter in lowercase:
            alphabet = lowercase
        elif letter in uppercase:
            alphabet = uppercase
        else:
            new_word += letter
            continue

        # Find the letter and shift its position around the alphabet.
        for position in range(26):
            if letter == alphabet[position]:
                new_position = (position + n) % 26
                new_word += alphabet[new_position]
                break

    return new_word

def acipher_ceaza(n, word):
    # Decryption moves the letters in the opposite direction.
    return cipher_ceaza(-n, word)

word = input("Enter a message (for example HELLO): ")
shift = int(input("Enter an integer shift (for example 3): "))
mode = input("Type encrypt or decrypt: ").lower()

if mode == "encrypt":
    print("Result:", cipher_ceaza(shift, word))
elif mode == "decrypt":
    print("Result:", acipher_ceaza(shift, word))
else:
    print("Please run the cell again and type encrypt or decrypt.")
```

```python
# Task6 (a)
# game of Hangman

import random
words = ["python", "java", "javascript"]

word = random.choice(words)
guessed_letters = []
incorrect_guesses = 0
max_guesses = 6

print("Welcome to Hangman!")

while incorrect_guesses < max_guesses:
    display_word = ""

    for letter in word:
        if letter in guessed_letters:
            display_word += letter + " "
        else:
            display_word += "_ "

    print("\nWord:", display_word)

    print("Incorrect guesses:", incorrect_guesses, "/", max_guesses)

    if all(letter in guessed_letters for letter in word):
        print("Congratulations! You won!")
        break

    guess = input("Guess a letter: ").lower()

    if len(guess) != 1 or not guess.isalpha():
        print("Please enter one letter.")
        continue

    if guess in guessed_letters:
        print("You already guessed that letter!")
        continue

    guessed_letters.append(guess)

    if guess in word:
        print("Correct guess!")
    else:
        incorrect_guesses += 1
        print("Wrong guess!")

if incorrect_guesses == max_guesses:
    print("Game over! The word was:", word)
```
